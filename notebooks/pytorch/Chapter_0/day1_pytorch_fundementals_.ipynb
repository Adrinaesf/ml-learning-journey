### Random Tensors

Why Random Tensors? 

Random Tensors are important because the way many neural network works is that they start with tensors full of random numbers and then adjust those numbers to better represent of data

`start with random num -> look at data -> update the random numbers -> look at data -> update random numbers`

```python
# Create a random tensor of size (3, 4)
random_tensor = torch.rand(3, 4, 10)
random_tensor
```

```python
random_tensor.ndim
```

```python
# Create a random tensor with a similar shape to an image tensor:
random_image_size_tensor = torch.rand(size=(224, 224, 3))
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

### Zeros and Ones

Helpful for creating a mask.
- Making some numbers to zero in your random tensor by multiplying them to zero tensor
- Saving some of the tensor info by multiplying it to ones. 

```python
# Create a tensor of all zeros: 
zeros = torch.zeros(3, 4)
zeros
```

```python
ones = torch.ones(3, 4)
ones
```

```python
ones.dtype
```

### Creating a range of tensors and tensors-like

```python
# Use torch.arange()
one_to_nine = torch.arange(start=1, end=10, step=1)
one_to_nine
```

```python
# Creating tensor-like use zeros_like, ones_like
nine_zeros = torch.zeros_like(one_to_nine)
nine_zeros
```
