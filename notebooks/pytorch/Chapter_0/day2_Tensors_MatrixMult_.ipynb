### 🔵 Tensor Data Type

General Knowledge: **Percisions**

    Precision of a numerical quantity that says how much detail the quantity expresses. 
    The default of torch.tensor is float32, which is 32 bits. 
    The lower the precision is, the faster calculations we have. 

**Note:** Tensor datatype is one of the 3 big errors we run into with PyTorch and DL: 
1. Tensors not right datatype --> ex: doing calculations with another tensor of different datatype
2. Tensors not correct shape. 
3. Tensors not on the correct device. --> Devices of tensors when doing calculations must be the same. 

```python
# Float 32 tensor
float_32_tensor = torch.tensor([3.0, 8.0, 9.0], 
                               dtype=None, # what datatype is the tensor (float32, float 16)
                               device=None, # what device is your tensor on, like gpu, cpu, cuba, ...
                               requires_grad=False) # whether or not to track gradients with these tensor operations. 
```

```python
float_32_tensor.dtype
```

```python
# Tackling the first problem:

float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
int_32_tensor = torch.tensor([1, 3, 6], dtype=torch.int32)
float_32_tensor * int_32_tensor
```

### 🔵 Manipulating Tensors (tensor operations)

Tensor Operations include: 
* Addition
* Subtraction
* Multiplication (element-wise)
* Division
* Matrix Multiplication

**Example:**

```python
tensor = torch.tensor([1, 2, 3])

print(tensor + 10)
print(tensor * 10)
print(tensor - 10)
print(tensor / 10)

# There are also pytorch built-in functions for operations: 
torch.mul(tensor, 10)
torch.add(tensor, 10)
```

### 🔵 Matrix Multiplication: 

Two main ways of performing in neural networks and DL

1. Element-wise multiplication
2. Matrix multiplication (dot product) --> row * col --> Use: `torch.matmul(tensor, tensor)`

More information on multiplying matrices: https://www.mathsisfun.com/algebra/matrix-multiplying.html


```python
# Element-wise multiplication: 
tensor = torch.tensor([1, 2, 3])

print(tensor, "*", tensor)
print(f"Equals: {tensor * tensor}")
```

```python
# Matrix Multiplication: 
torch.matmul(tensor, tensor)
'''
[1, 2, 3] * [1] = 1*1 + 2*2 + 3*3 = 14
            [2]
            [3]
'''
# Matrix mult by code: 
value = 0
for i in range(len(tensor)):
    value += tensor[i] * tensor[i]
print(value)

# It is way faster with torch.matmul(tensor, tensor)
```

### 🔵 One of the most common error in DL:

There are two main rules that performing matrix multipication needs to satisfy: 
1. The **inner dimensions** must match:
* `(3, 2) @ (3, 2)` won't work
* `(2, 3) @ (3, 2)` will work
* `(3, 2) @ (2, 3)` will work

2. The resulting matrix has the shape of the **outer dimensions:**
* `(2, 3) @ (3, 2) -> (2, 2)`
* `(3, 10) @ (10, 2) -> (3, 2)`

```python
# Shape for matrix multiplication
tensor_A = torch.tensor([[1, 2], 
                         [3, 4], 
                         [5, 6]])

tensor_B = torch.tensor([[7, 10], 
                         [8, 11], 
                         [9, 12]])

# torch.mm(tensor_A, tensor_B) #torch.mm = torch.matmul
# tensor_A = (3x2)
# tensor_B = (3x2) 
# Problem!! 
```

To fix our tensor shape issues, we can manipulate the shape of one of our tensors using a **transpose**. 
A **transpose** switches the axes or dimensions of a given tensor. 
How to do it? `tensorName.T`

```python
print(tensor_B, tensor_B.shape)
print("Now we transpose")
print(tensor_B.T, tensor_B.T.shape)
```

```python
# The matrix multiplication operation works when tensor_B is transposed. 
print(f"Original shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}")
print(f"New shapes: tensor_A = {tensor_A.shape} (same as above), tensor_B.T = {tensor_B.T.shape}")
print(f"Multiplying: {tensor_A.shape} @ {tensor_B.T.shape} <- inner dimensions must match")
print("Output:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
print(f"\nOutput shape: {output.shape}")
```
