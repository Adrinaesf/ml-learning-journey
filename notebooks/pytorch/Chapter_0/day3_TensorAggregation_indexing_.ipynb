## 🔵 Finding the min, max, mean, sum (tensor aggregation)
- Finding the min: `torch.min(tensorName)/x.min()`
- Finding the max: `torch.max(tensorName)/x.max()`
- Finding the sum: `torch.sum(tensorName)/x.sum()`
- Finding the mean: we need to have tensor of type float32
  `torch.mean(x.type(torch.float32))`

```python
# Create a tensor
x = torch.arange(1, 100, 10)
print(x)

# Finding the min: 
print(f"The min value is: {torch.min(x)}")
print(f"The max value is: {x.max()}")
print(f"The sum value is: {torch.sum(x)}")
print(f"The mean value is: {torch.mean(x.type(torch.float32))}")
```

## 🔵 Finding the position of min, max, ...
- Finding the position of min: `tensorName.argmin()`
- Finding the position of max: `tensorName.argmax()`

```python
print(x)
print(f"The index/position of min value is: {x.argmin()}")
print(f"The index/position of max value is: {x.argmax()}")
```

## 🔵 Reshaping, stacking, squeezing, and unsqueezing

**Useful in helping us debugging the shape and dimensions essiues**

- **Reshaping**: reshapes an input tensor to a defined shape
- **View**: Return a view of an input tensor of certain shape but keep the same memory as the original tensor
- **Stacking**: Compbine multiply tenosr on top of each other (vstack) or side by side (hstack)
- **Squeeze**: Remove all `1` dimensions from a tensor
- **Unsqueeze**: add all `1` dimensions to a target tensor
- **Permute**: Return a view of the input with dimensions permuted (swapped) in a certain way


#### 🟢 Reshaping and View:

```python
# Let's create a tensor first
x = torch.arange(1., 10.)
x, x.shape
```

```python
# Add an extera dimensions
x_reshaped = x.reshape(3, 3)
x_reshaped, x_reshaped.shape
```

```python
# Change the view: A view return a view of input tensor, but their memories are same
z = x.view(1,9)
z, z.shape

# change in z = change in x
z[:, 0] = 5
z, x
```

#### 🟢 Stack and Squeeze:

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x], dim=0)
x_stacked
# The dim should be compatible to the actual tensor x. 
# similar thing for adding a dimension
```

```python
x_reshaped = x.reshape(1, 9)
# Squeeze and Unsqueeze
# torch.squeeze - removes all single dimensions from target tensor

print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

# Now we romve the extera dimension from x_reshaped:
x_reshaped_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_reshaped_squeezed}")
print(f"New shape: {x_reshaped_squeezed.shape}")
```

```python
# Another example
y = torch.tensor([[[1, 2, 3]]])
y_reshaped = y.reshape(1, 3, 1)

print(f"Previous tensor: {y_reshaped}")
print(f"Previous shape: {y_reshaped.shape}")

# Now we romve the extera dimensions from y_reshaped:
y_reshaped_squeezed = y_reshaped.squeeze()
print(f"\nNew tensor: {y_reshaped_squeezed}")
print(f"New shape: {y_reshaped_squeezed.shape}")
```

#### 🟢 Unsqueeze:

```python
# torch unsqueeze - adds a single dimension ro a target tensor at a specific dim
print(f"Previous squeezed shape: {y_reshaped_squeezed}")
print(f"Previous squeezed shape: {y_reshaped_squeezed.shape}")

# Add an extera dimension with unsqueeze:
y_unsqueezed = y_reshaped_squeezed.unsqueeze(dim=1)
print(f"\nNew tensor: {y_unsqueezed}")
print(f"New shape: {y_unsqueezed.shape}")
```

#### 🟢 Permute:

```python
# Permute basically changes the dimensions. 
# In ML dimensions represent a sort of information of our data
# Like in photos we have height width, and RGB
# Now let's say we have a picture that is beinge represented by a tensor of numbers
# and its shape is like size=(224, 224, 3) (height, width, RGB)
# Now we want to have the shape of (3, 224, 224) (RGB, height, width)
# So the RGB is at the beginning

# we map using permute:
# map perv index -> new index
# map index(2) -> index(0) so RGB at dim=0
# map index(0) -> index(1) so height at dim=1
# map index(1) -> index(2) so width at dim=2

# Permute(index(2), index(0), index(1)) so index 2 at the index 0 of permute
# index(0) at the index(1) of permute and continuing to fill the arguments...

x_original = torch.rand(size=(224, 224, 3)) # [height, width, colour_channel]

# permute the original tensor to rearrange the axis (or dim) order
x_permuted = x_original.permute(2, 0, 1)

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
```

🟣 **Challenge:** Change a value of x_original, and see if it changes in x_permuted. If so, explain why?

```python
print(f"Previous value at index 0, 0, 0 of x_original: {x_original[0, 0, 0]}")
x_original[0, 0, 0] = 10

print(f"New value at index 0, 0, 0 of x_original: {x_original[0, 0, 0]}")
# Now we check the x_permuted:
print(f"New value at index 0, 0, 0 of x_permuted: {x_permuted[0, 0, 0]}")
```

_Answer: The reason is that x_permuted is a view of x_original, and a view has the same memory address of x_original. So any changes in x_original will result a change in x_permuted as well._

## 🔵 Selecting Data from Tensors (Indexing)

Indexing in PyTorch is similar to indexing with Numpy. 

```python
# Create a Tensor:
x = torch.arange(1, 10).reshape(1, 3, 3)
x, x.shape
```

```python
# Let's index on our new tensor: 
x[0]
```

```python
x[0][0]
```

```python
x[0][0][0]
```

```python
x[0, 2, 2]
```

```python
x[:, :, 1]
```

```python
# Get all values of 0th and 1st dimensions but only index 1 of second dimension
x[:, :, 1]
# The indexing is related to the dimesnions. 
# So when we have : in dim=0, it goes through all elements in the most outer []
# Similarly : in dim=1 goes through all the rows we have. 
# but 1 in dim=2 means that we only want the value at index 1 of dim2, so in each row
# we take the second col/index
```

```python
# Get all values of the 0 dimension but only the 1 index of 1 and second dimension:
x[:, 1, 1]
```

```python
# Get index 0 of 0th and 1st dimension and all values of 2nd dimension: (first row)
print(x[0, 0, :])
# get index 0 of 0th and 2nd dimension and all values of first dimension: (first col)
print(x[0, :, 0])
```

## 🔵 PyTorch and NumPy

NumPy is a popular scientific Python numerical computing library. 
And because of this, PyTorch has functionality to interact with it. 

* NumPy --> PyTorch tensor: `torch.from_numpy(ndarray)`
* PyTorch tensor --> NumPy: `torch.Tensor.numpy()`

```python
# NumPy array to Tensor:

array = np.arange(1.0, 8.0)
# Warning: when converting from numpy -> pytorch, pytorch reflects numpy a 
# default of dtype = float64

# Tensor will get a new memory:
tensor = torch.from_numpy(array) 
array, tensor
```

```python
# Change the value of the array?
array_changed = array + 1
array_changed, tensor
```

```python
# Tensor to numpy array: 
tensor_example = torch.ones(7)
numpy_tensor = tensor_example.numpy()
tensor_example, numpy_tensor
```

```python
tensor_changed = tensor_example + 1
tensor_changed, numpy_tensor
```
